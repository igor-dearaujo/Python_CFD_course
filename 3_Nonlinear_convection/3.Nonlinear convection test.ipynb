```python
import numpy as np
from matplotlib import pyplot as plt
```

```python
domain = np.array([0, 2])
nx = 200
dx = (domain[1] - domain[0])/ (nx - 1)

x = np.linspace(domain[0], domain[1], nx)

#initialize u function
u = np.ones(nx)
u[int(0.5/ dx) : int(1/dx + 1)] = 2 #create hat function
u0 = u.copy()

print('Plotting initial conditions...')
plt.plot(x, u)
plt.title('Initial conditions')

nt = 20 #number of time steps solved
CFL = 1
dt = CFL * dx / max(abs(u))


print(f'Plotting {nt} iterations of the solution ...')
for n in range(nt):
    un = u.copy() #create un to stores values from previous iterations

    for i in range(1, nx-1):

        if un[i] >= 0:
            #backward scheme
            u[i] = un[i] - un[i] * (dt/dx) * (un[i] - un[i-1])
        else:
            #forward scheme
            u[i] = un[i] - un[i] * (dt/dx) * (un[i+1] - un[i])

    plt.plot(x, u)
    plt.plot(x, u0)
    plt.show()
```

```python
#now with negative initial conditions
#initialize u function
u = np.ones(nx)
u[:] = -1
u[int(0.5/ dx) : int(1/dx + 1)] = -2 #create hat function
u0 = u.copy()

print('Plotting initial conditions...')
plt.plot(x, u)
plt.title('Initial conditions')

print(f'Plotting {nt} iterations of the solution ...')
for n in range(nt):
    un = u.copy() #create un to stores values from previous iterations

    for i in range(1, nx-1):

        if un[i] > 0:
            #backward scheme
            u[i] = un[i] - un[i] * (dt/dx) * (un[i] - un[i-1])
        else:
            #forward scheme
            u[i] = un[i] - un[i] * (dt/dx) * (un[i+1] - un[i])

    plt.plot(x, u)
    plt.plot(x, u0)
    plt.title(f'U print in {n} timestep')
    plt.show()
```
